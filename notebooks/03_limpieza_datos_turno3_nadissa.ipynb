# Limpieza de datos - turno 3: Nadissa

Este notebook ejecuta las reglas aprobadas para `NIVEL`, `SECTOR`, `AREA`, `STATUS`, `MODALIDAD`, `JORNADA` y `PLAN`. Las reglas viven en `src/limpieza.py`; aqui solo se presentan resultados y se genera la salida preliminar reproducible. 

```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import display

ROOT = Path('..').resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.diagnostico import cargar_datos
from src.limpieza import CATEGORIAS_NADISSA, limpiar_datos_preliminar
from src.validacion import validar_categorias_nadissa

crudo = cargar_datos()
limpio_pre = limpiar_datos_preliminar(crudo)
validar_categorias_nadissa(limpio_pre)
print(f'Registros conservados: {len(limpio_pre):,}')
```

## Resultados de las variables asignadas

`AREA = SIN ESPECIFICAR` se convierte a `NA`. Los doce registros de Ciudad Capital con `AREA = RURAL` se conservan. Nadissa verifico manualmente los codigos `00-01-0512-46`, `00-02-0137-46` y `00-07-0007-46`: codigo, nombre y direccion coinciden exactamente con la fuente; los tres pertenecen al area rural.

```python
resumen = pd.DataFrame([
    {'variable': columna, 'tipo_final': str(limpio_pre[columna].dtype),
     'dominio': ', '.join(dominio), 'faltantes': int(limpio_pre[columna].isna().sum())}
    for columna, dominio in CATEGORIAS_NADISSA.items()
])
display(resumen)

capital_rural = limpio_pre[
    limpio_pre['DEPARTAMENTO_ORIGINAL'].eq('CIUDAD CAPITAL')
    & limpio_pre['AREA'].eq('RURAL')
][['CODIGO', 'ZONA_CAPITAL', 'ESTABLECIMIENTO', 'DIRECCION', 'AREA']]
display(capital_rural)
```

## Salida preliminar

La salida conserva las 11,890 filas y todas las columnas de trazabilidad generadas por los turnos anteriores. Se escribe en `data/interim/`, nunca en `data/raw/`.

```python
salida = ROOT / 'data' / 'interim' / 'establecimientos_diversificado_limpio_preliminar.csv'
limpio_pre.to_csv(salida, index=False, encoding='utf-8-sig')
print(f'CSV generado: {salida}')
```
